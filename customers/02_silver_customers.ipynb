```python
from pyspark.sql.functions import col, to_date

df_bronze = spark.read.table("workspace.bronze.bronze_customers")

display(df_bronze)

df_silver = (
    df_bronze
    .drop("index")
    .withColumn("subscription_date", to_date(col("subscription_date"), "yyyy-MM-dd"))
    .dropDuplicates(["customer_id"])
)

display(df_silver)

df_silver.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.bronze.silver_customers")
```
