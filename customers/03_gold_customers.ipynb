```python
from pyspark.sql.functions import col

df_silver = spark.read.table("workspace.silver.silver_customers")

df_gold = (
    df_silver
    .groupBy("country")
    .count()
    .orderBy(col("count").desc())
)

display(df_gold)
```

```python
df_gold.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.gold.gold_customers_by_country")
```

```python
from pyspark.sql.functions import col, date_format

df_silver = spark.read.table("workspace.silver.silver_customers")

df_gold_month = (
    df_silver
    .withColumn("subscription_month", date_format(col("subscription_date"), "yyyy-MM"))
    .groupBy("subscription_month")
    .count()
    .orderBy("subscription_month")
)

display(df_gold_month)
```

```python
df_gold_month.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.gold.gold_customers_by_subscription_month")
```
