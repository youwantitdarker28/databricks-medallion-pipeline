```python
# Run once to create the audit schema + table
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.audit")

spark.sql("""
    CREATE TABLE IF NOT EXISTS workspace.audit.people_pipeline_runs (
        run_id          STRING,
        pipeline        STRING,
        layer           STRING,
        start_ts        TIMESTAMP,
        end_ts          TIMESTAMP,
        status          STRING,
        watermark_in    TIMESTAMP,
        watermark_out   TIMESTAMP,
        rows_read_bronze LONG,
        rows_upserted_silver LONG,
        error_message   STRING
    )
    USING DELTA
""")

print("Audit table ready.")
```
